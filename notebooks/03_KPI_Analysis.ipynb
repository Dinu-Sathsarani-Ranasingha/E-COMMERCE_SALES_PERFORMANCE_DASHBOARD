# Phase 3 – KPI Definition & Business Insights

## Objective

The objective of this phase is to calculate key business performance indicators (KPIs), identify meaningful business insights, and prepare the data for the Power BI dashboard.

### Tools Used

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn

### KPIs

- Total Revenue
- Total Profit
- Profit Margin
- Average Order Value
- Average Discount

```python
# KPI 1 - Total Revenue

total_revenue = df["Sales"].sum()

print(f"Total Revenue: ${total_revenue:,.2f}")
```

## KPI 1 – Total Revenue

**Definition**

Total Revenue represents the total sales generated by the business during the analysis period.

**Formula**

Total Revenue = Sum of Sales

**Business Value**

This KPI measures the overall sales performance of the business and serves as a key indicator of business growth.

```python
# KPI 2 - Total Profit

total_profit = df["Profit"].sum()

print(f"Total Profit: ${total_profit:,.2f}")
```

## KPI 2 – Total Profit

**Definition**

Total Profit represents the total earnings after subtracting all business costs.

**Formula**

Total Profit = Sum of Profit

**Business Value**

This KPI shows whether the business is generating sustainable earnings in addition to revenue.

```python
# KPI 3 - Profit Margin

profit_margin = (total_profit / total_revenue) * 100

print(f"Profit Margin: {profit_margin:.2f}%")
```

## KPI 3 – Profit Margin

**Definition**

Profit Margin measures the percentage of revenue that is converted into profit.

**Formula**

Profit Margin = (Total Profit ÷ Total Revenue) × 100

**Business Value**

A higher profit margin indicates better operational efficiency and profitability.

```python
# KPI 4 - Average Order Value (AOV)

total_orders = df["Order ID"].nunique()

average_order_value = total_revenue / total_orders

print(f"Total Orders: {total_orders:,}")
print(f"Average Order Value (AOV): ${average_order_value:,.2f}")
```

## KPI 4 – Average Order Value (AOV)

**Definition**

Average Order Value (AOV) measures the average revenue generated per customer order.

**Formula**

Average Order Value = Total Revenue ÷ Total Number of Orders

**Business Value**

A higher AOV indicates that customers spend more per transaction, helping increase revenue without necessarily increasing the number of orders.

```python
# KPI 5 - Average Discount

average_discount = df["Discount"].mean() * 100

print(f"Average Discount: {average_discount:.2f}%")
```

## KPI 5 – Average Discount

**Definition**

Average Discount represents the average percentage discount offered across all orders.

**Formula**

Average Discount = Mean of Discount × 100

**Business Value**

Monitoring discount levels helps businesses balance sales growth with profitability. Excessive discounts may reduce overall profit margins.

```python
# KPI Summary Table

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Profit",
        "Profit Margin",
        "Average Order Value",
        "Average Discount"
    ],
    "Value": [
        f"${total_revenue:,.2f}",
        f"${total_profit:,.2f}",
        f"{profit_margin:.2f}%",
        f"${average_order_value:,.2f}",
        f"{average_discount:.2f}%"
    ]
})

kpi_summary
```

```python
monthly_sales = df.groupby(["Year", "Month"])["Sales"].sum().reset_index()

monthly_sales["Period"] = pd.to_datetime(
    monthly_sales[["Year", "Month"]].assign(day=1)
)

plt.figure(figsize=(12,5))

plt.plot(
    monthly_sales["Period"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Date")
plt.ylabel("Revenue ($)")

plt.tight_layout()
plt.show()
```

```python
profit_category = df.groupby("Category")["Profit"].sum().sort_values()

plt.figure(figsize=(8,5))

profit_category.plot(kind="bar")

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit ($)")

plt.tight_layout()
plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values()

plt.figure(figsize=(8,5))

segment_sales.plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales ($)")

plt.tight_layout()
plt.show()
```

### Sales by Customer Segment

This chart compares total sales across different customer segments. It helps identify which customer group contributes the most revenue and supports targeted marketing and customer engagement strategies.

```python
plt.figure(figsize=(10,5))

sns.histplot(df["Discount"], bins=20, kde=True)

plt.title("Distribution of Discounts")
plt.xlabel("Discount")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()
```

### Discount Distribution

This chart shows how discounts are distributed across customer orders. Understanding discount patterns helps evaluate pricing strategies and their potential impact on profitability.

```python
plt.figure(figsize=(8,6))

corr = df[["Sales","Profit","Quantity","Discount"]].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="Blues",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

### Correlation Analysis

The heatmap illustrates the relationships between key numerical variables such as Sales, Profit, Quantity, and Discount. It helps identify positive and negative correlations that can support business decision-making.

```python
segment_profit = (
    df.groupby("Segment")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

segment_profit
```

```python
plt.figure(figsize=(8,5))

plt.bar(
    segment_profit["Segment"],
    segment_profit["Profit"]
)

plt.title("Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Profit ($)")

plt.tight_layout()
plt.show()
```

### Customer Segment Analysis

The Consumer segment generated the highest total profit, making it the most valuable customer segment for the business. The Corporate segment ranked second, while the Home Office segment contributed the lowest profit. These findings suggest that customer retention and marketing efforts should primarily focus on the Consumer segment while exploring strategies to improve profitability within the Home Office segment.
